```python
import numpy as np
import matplotlib.pyplot as plt
import pymcel as pc
```

Trayectoria del vector relativo en el sistema perifocal

```python
p = 1
e = 0.7

thetaer = np.linspace(0, 2 * np.pi, 100)
rs = p / (1 + e * np.cos(thetaer))

xfs = rs * np.cos(thetaer)
yfs = rs * np.sin(thetaer)
```

```python
# Graph!! 
plt.plot(xfs, yfs)
plt.grid()
plt.axis('equal')
plt.show()
```

```python
from ipywidgets import interact
```

```python
# Interactive code for eccentricity

@interact(e=(0, 1, 0.001))
def draw_orbit(e):
    rs = p / (1 + e * np.cos(thetaer))

    xfs = rs * np.cos(thetaer)
    yfs = rs * np.sin(thetaer)
    
    plt.figure(figsize=(5, 5))
    plt.plot(xfs, yfs)
    plt.xlim(-p* 2, p * 2)
    plt.ylim(-p* 2, p * 2)
    plt.grid()
    plt.title(f'Eccentricity: {e:.2f}')
    plt.show()
```

### Jacobi Transform

```python
p = 1
e = 0.7

thetaer = np.linspace(0, 2 * np.pi, 100)
rs = p / (1 + e * np.cos(thetaer))

xfs = rs * np.cos(thetaer)
yfs = rs * np.sin(thetaer)
```

```python
# Trayectory per particle

m1 = 1
m2 = 0.7
M = m1 + m2
RCM = m1 + m2

rvs = np.array([xfs, yfs])
rvs = rvs.T

r1v = RCM + m2/M * rvs
r2v = RCM - m1/M * rvs

plt.plot(r1v[:,0], r1v[:,1], label='Particle 1')
plt.plot(r2v[:,0], r2v[:,1], label='Particle 2')
#plt.plot(xfs, yfs, label='Center of Mass', linestyle='--')
plt.grid()
plt.axis('equal')
plt.legend()
plt.show()
```
